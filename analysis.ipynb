```python
from pathlib import Path
from tflmfi import generate_physical_report
from IPython.display import display

# Caminho da campanha
run_dir = Path("runs/Memoria_transient")

# Gera relatório físico
report = generate_physical_report(
    run_dir=run_dir,
    bin_size=128,        # ajuste se quiser bins mais finos/grossos
    top_n_regions=10
)

# Resumo em texto
print(report.summary())

# Salva os artefatos na própria pasta da campanha
saved_files = report.save(run_dir)
print("\nArquivos gerados:")
for p in saved_files:
    print("-", p)

# Visualizações rápidas no notebook
print("\nTop 10 tensores mais críticos:")
display(report.by_tensor.head(10))

print("Top 10 regiões físicas mais críticas:")
display(report.critical_regions.head(10))

print("Criticidade por operador (top 10):")
display(report.by_operator.head(10))
```
